# 📊 07. Machine Learning Model Evaluation

This notebook implements predictions using the deployed best model (`best_model.pkl`), evaluates all standard metric definitions (Accuracy, F1-Score, ROC-AUC), plots confusion matrices and ROC curves, and reviews business relevance.

## 1. Introduction & Setup

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Append project root
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))

from src.models.predict import RiskPredictor
from configs.config import config
from configs.constants import TARGET_COL

print("Evaluation packages imported successfully.")
```

## 2. Load Deployed Best Model & Test Data

```python
paths = config.get_paths()
X_test = pd.read_csv(os.path.join(paths['processed_dir'], "X_test.csv"))
y_test = pd.read_csv(os.path.join(paths['processed_dir'], "y_test.csv"))[TARGET_COL]

predictor = RiskPredictor()
print(f"Loaded test split with shape: {X_test.shape}")
```

## 3. Generate Predictions & Probabilities

```python
y_pred = predictor.predict(X_test)
y_prob = predictor.predict_probability(X_test)

print("Sample predictions:", y_pred[:10])
print("Sample probabilities:", [round(p, 4) for p in y_prob[:10]])
```

## 4. Evaluation Metrics Summary

```python
from src.models.metrics import calculate_all_metrics

metrics = calculate_all_metrics(y_test, np.array(y_pred), np.array(y_prob))
for k, v in metrics.items():
    print(f"{k}: {v:.4f}")
```
